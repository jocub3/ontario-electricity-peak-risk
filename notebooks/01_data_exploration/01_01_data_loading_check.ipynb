# 01_01 - Data Loading Check

```python
import pandas as pd
from pathlib import Path

DATA_DIR = Path.cwd().parents[1] / "data" / "processed"
DATA_DIR
```

## 1. Calendar

One row per hour, with date and time features already built in (season, weekday, holiday flags, cyclical encodings). Join key: `timestamp_local`, a fixed-clock timestamp with no daylight saving shifts.

```python
calendar_df = pd.read_csv(DATA_DIR / "calendar_hourly_ontario_timestamp.csv")

print("shape:", calendar_df.shape)
calendar_df.info()
```

```python
calendar_df.head()
```

## 2. Weather - Toronto City

Covers the downtown FSAs (M5S, M5R, M6G). Reports temperature, humidity and precipitation, not wind speed or visibility. Join key: `Date/Time (LST)`.

```python
weather_city_df = pd.read_csv(DATA_DIR / "weather_toronto_city_6158355.csv")

print("shape:", weather_city_df.shape)
weather_city_df.info()
```

```python
weather_city_df.head()
```

## 3. Weather - Toronto INTL A (Pearson)

Covers the airport-area FSAs (L4T, M9W, M9R). Reports wind speed and visibility, not precipitation. Join key: `Date/Time (LST)`.

```python
weather_intl_df = pd.read_csv(DATA_DIR / "weather_toronto_intl_a_6158731.csv")

print("shape:", weather_intl_df.shape)
weather_intl_df.info()
```

```python
weather_intl_df.head()
```

## 4. Hourly Consumption per FSA (6 files)

One file per FSA, join key `TIMESTAMP`. Unlike calendar and weather, this is not one row per hour: it is one row per (FSA, hour, customer type, price plan), so the same hour appears multiple times, split across customer segments.

```python
FSAS = ["l4t", "m5r", "m5s", "m6g", "m9r", "m9w"]

consumption_dfs = {}
load_summary = []
for fsa in FSAS:
    df = pd.read_csv(DATA_DIR / f"hourly_consumption_fsa_{fsa}.csv")
    consumption_dfs[fsa.upper()] = df
    load_summary.append({"FSA": fsa.upper(), "rows": df.shape[0], "cols": df.shape[1], "FSA values": df["FSA"].unique()})

pd.DataFrame(load_summary)
```

```python
# Using M5S as a representative sample - all six files share the same structure
consumption_dfs["M5S"].info()
```

```python
consumption_dfs["M5S"].head()
```

## 5. Comparing the datasets side by side

This section looks at how the files relate to each other: whether they cover the same period of time, and how much of the row count in the consumption files comes from real hourly variation versus the customer-type/price-plan split.

```python
date_ranges = [
    {"dataset": "calendar", "min": calendar_df["timestamp_local"].min(), "max": calendar_df["timestamp_local"].max()},
    {"dataset": "weather_toronto_city", "min": weather_city_df["Date/Time (LST)"].min(), "max": weather_city_df["Date/Time (LST)"].max()},
    {"dataset": "weather_toronto_intl_a", "min": weather_intl_df["Date/Time (LST)"].min(), "max": weather_intl_df["Date/Time (LST)"].max()},
]
date_ranges += [
    {"dataset": f"consumption_{fsa}", "min": df["TIMESTAMP"].min(), "max": df["TIMESTAMP"].max()}
    for fsa, df in consumption_dfs.items()
]

pd.DataFrame(date_ranges)
```

The calendar and weather files run through the end of 2026, while consumption stops at the end of 2025. The smart-meter readings simply have not been produced for those later dates yet. Since consumption defines what can actually be modeled, 2021-2025 is the real usable window, and the 2026 rows in calendar and weather need to be left out once everything is joined.

```python
# CUSTOMER_TYPE / PRICE_PLAN combinations per FSA, explains why there are multiple rows per FSA-hour
combos_by_fsa = pd.concat(
    [
        df[["CUSTOMER_TYPE", "PRICE_PLAN"]].drop_duplicates().assign(FSA=fsa)
        for fsa, df in consumption_dfs.items()
    ],
    ignore_index=True,
)[["FSA", "CUSTOMER_TYPE", "PRICE_PLAN"]].sort_values(["FSA", "CUSTOMER_TYPE", "PRICE_PLAN"])

combos_by_fsa
```

```python
# Final summary: shape of each raw dataset
summary = {
    "calendar": calendar_df.shape,
    "weather_toronto_city": weather_city_df.shape,
    "weather_toronto_intl_a": weather_intl_df.shape,
    **{f"consumption_{fsa}": df.shape for fsa, df in consumption_dfs.items()},
}
pd.Series(summary, name="shape (rows, cols)")
```

The combination counts above explain the row-count gap seen earlier between, for example, M5R (about 101K rows) and L4T (about 216K rows): every FSA is still one row per hour, just multiplied by however many customer-type/price-plan segments exist for that FSA. This needs to be collapsed to a single row per FSA-hour before the data can be used for modeling.
