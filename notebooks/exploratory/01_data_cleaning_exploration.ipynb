```python
import pandas as pd
import numpy as np
from pathlib import Path

pd.set_option("display.max_columns", None)
```

```python
DATA_PATH = Path("../data/processed")

consumption = pd.read_csv(
    DATA_PATH / "hourly_consumption_fsa_l4t.csv"
)

consumption.head()
```

```python
print("Shape:", consumption.shape)
consumption.info()
```

```python
consumption.isna().sum().sort_values(ascending=False)
```

```python
# Basic data integrity checks

print("Exact duplicate rows:", consumption.duplicated().sum())

print("\nFSA values:")
print(consumption["FSA"].value_counts())

print("\nCustomer types:")
print(consumption["CUSTOMER_TYPE"].value_counts())

print("\nPrice plans:")
print(consumption["PRICE_PLAN"].value_counts())

print("\nTOTAL_CONSUMPTION range:")
print("Min:", consumption["TOTAL_CONSUMPTION"].min())
print("Max:", consumption["TOTAL_CONSUMPTION"].max())

print("\nPREMISE_COUNT range:")
print("Min:", consumption["PREMISE_COUNT"].min())
print("Max:", consumption["PREMISE_COUNT"].max())

print("\nHOUR values:")
print(sorted(consumption["HOUR"].unique()))
```
