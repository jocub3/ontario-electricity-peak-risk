## Interpretation Note

Variance Inflation Factor (VIF) is used only as a diagnostic indicator in this project.

High or infinite VIF values do not automatically result in feature removal because several engineered predictors are mathematically related by design, including cyclical encodings, DST indicators, temperature-derived variables, lagged demand variables, and rolling statistics.

Final multicollinearity treatment will be model-specific. It will be considered primarily for linear and statistical models, while tree-based models such as Random Forest, XGBoost, and LightGBM will not use VIF as an automatic exclusion criterion.
