# DSA_07 — Deployment Validation

**Purpose.** Verify that the public application package is lightweight and
does not require the serialized model artifacts.

```python
# Import libraries
from pathlib import Path
import yaml
```

```python
# Define project/config paths
PROJECT_ROOT = Path.cwd()
while not (PROJECT_ROOT / "src").exists() and PROJECT_ROOT != PROJECT_ROOT.parent:
    PROJECT_ROOT = PROJECT_ROOT.parent

cfg = yaml.safe_load(
    (PROJECT_ROOT / "configs" / "decision_support_app.yaml")
    .read_text(encoding="utf-8")
)
```

```python
# Validate available data
required = [
    PROJECT_ROOT / cfg["paths"]["demo_predictions"],
    PROJECT_ROOT / cfg["paths"]["demo_metadata"],
    PROJECT_ROOT / cfg["deployment"]["app_entrypoint"],
    PROJECT_ROOT / cfg["deployment"]["requirements_file"],
]

missing = [str(p) for p in required if not p.exists()]
if missing:
    raise FileNotFoundError(
        "Deployment prerequisites are missing:\n" + "\n".join(missing)
    )

assert cfg["deployment"]["load_model_artifacts"] is False
assert cfg["deployment"]["include_large_models_in_repo"] is False

print("Deployment artifact validation: PASS")
print("Public model loading: DISABLED")
```

```python
data_files = [
    PROJECT_ROOT / cfg["paths"]["demo_predictions"],
    PROJECT_ROOT / cfg["paths"]["demo_metadata"],
]
scenario_path = PROJECT_ROOT / cfg["paths"]["demo_weather_scenarios"]
if scenario_path.exists():
    data_files.append(scenario_path)

size_mb = sum(p.stat().st_size for p in data_files) / (1024**2)
print(f"Precomputed public data size: {size_mb:.2f} MB")
print("DSA_07 RESULT: PASS")
```
