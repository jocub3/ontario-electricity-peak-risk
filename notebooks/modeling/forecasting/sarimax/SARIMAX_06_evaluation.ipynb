```python
# Import libraries
from pathlib import Path
import sys
import json
import pandas as pd
```

```python
# Define the root directory of the project and config file
PROJECT_ROOT = Path.cwd()
while not (PROJECT_ROOT / "src").exists() and PROJECT_ROOT != PROJECT_ROOT.parent:
    PROJECT_ROOT = PROJECT_ROOT.parent

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

REPORTS_DIR = PROJECT_ROOT / "reports" / "modeling" / "forecasting" / "sarimax"
OUTPUTS_DIR = PROJECT_ROOT / "outputs" / "modeling" / "forecasting" / "sarimax"

REPORTS_DIR, OUTPUTS_DIR
```

```python
import matplotlib.pyplot as plt
global_metrics = pd.read_csv(REPORTS_DIR / "SARIMAX_06_global_metrics.csv")
fold_metrics = pd.read_csv(REPORTS_DIR / "SARIMAX_06_fold_metrics.csv")
fsa_metrics = pd.read_csv(REPORTS_DIR / "SARIMAX_06_fsa_metrics.csv")
horizon_metrics = pd.read_csv(REPORTS_DIR / "SARIMAX_06_horizon_metrics.csv")
predictions = pd.read_parquet(OUTPUTS_DIR / "sarimax_validation_predictions.parquet")
predictions["forecast_origin"] = pd.to_datetime(predictions["forecast_origin"])
```

```python
for fold, group in horizon_metrics.groupby("fold"):
    plt.figure(figsize=(10, 5))
    plt.plot(group["horizon"], group["mae"], marker="o")
    plt.xlabel("Forecast Horizon (hours)")
    plt.ylabel("MAE (kWh)")
    plt.title(f"SARIMAX — MAE by Forecast Horizon ({fold})")
    plt.xticks(range(1, 25))
    plt.grid(alpha=0.25)
    plt.show()
```

```python
sample = predictions.sample(n=min(10000, len(predictions)), random_state=42)
plt.figure(figsize=(7, 7))
plt.scatter(sample["actual"], sample["predicted"], alpha=0.15)
minimum = min(sample["actual"].min(), sample["predicted"].min())
maximum = max(sample["actual"].max(), sample["predicted"].max())
plt.plot([minimum, maximum], [minimum, maximum], linestyle="--")
plt.xlabel("Actual Consumption (kWh)")
plt.ylabel("Predicted Consumption (kWh)")
plt.title("SARIMAX — Actual vs Predicted")
plt.grid(alpha=0.2)
plt.show()
```
