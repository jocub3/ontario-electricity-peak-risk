# RFFR 01 — Input Validation (Corrected)

Validate the exact functions exposed by the existing Random Forest candidate.

```python
from pathlib import Path
import sys
import yaml

PROJECT_ROOT = Path.cwd()
while not (PROJECT_ROOT / "src").exists() and PROJECT_ROOT != PROJECT_ROOT.parent:
    PROJECT_ROOT = PROJECT_ROOT.parent

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

CONFIG_PATH = PROJECT_ROOT / "configs" / "random_forest_final_refinement.yaml"
```

```python
from src.ontario_peak_risk.models.forecasting.random_forest_final_refinement.adapter import (
    candidate_capabilities,
)

with CONFIG_PATH.open("r", encoding="utf-8") as handle:
    CONFIG = yaml.safe_load(handle)

original_config = PROJECT_ROOT / CONFIG["paths"]["original_model_config"]

print("Original config:", original_config)
print("Exists:", original_config.exists())

capabilities = candidate_capabilities()
display(capabilities)

assert original_config.exists(), "Original Random Forest YAML was not found."
assert capabilities.iloc[0]["import_ok"], "Original Random Forest module could not be imported."
assert capabilities.iloc[0]["has_tune"], "Expected function `_tune_parameters` was not found."
assert capabilities.iloc[0]["has_run_model"], "Expected function `run_random_forest_regressor` was not found."

print("\nValidation passed. You can proceed to RFFR_02_quick_refinement.ipynb.")
```
