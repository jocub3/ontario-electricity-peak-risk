# XGBFR 01 — Input Validation

Verify that the selected XGBoost validation predictions and frozen development artifacts are available.

```python
# Import libraries
from pathlib import Path
import sys
import yaml
import pandas as pd
```

```python
# Define config paths

PROJECT_ROOT = Path.cwd()
while not (PROJECT_ROOT / "src").exists() and PROJECT_ROOT != PROJECT_ROOT.parent:
    PROJECT_ROOT = PROJECT_ROOT.parent

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

CONFIG_PATH = (
    PROJECT_ROOT
    / "configs"
    / "peak_risk_final_refinement.yaml"
)
CONFIG_PATH
```

```python


with CONFIG_PATH.open("r", encoding="utf-8") as handle:
    CONFIG = yaml.safe_load(handle)

prediction_path = PROJECT_ROOT / CONFIG["paths"]["validation_predictions"]
best_parameters_path = PROJECT_ROOT / CONFIG["paths"]["best_parameters"]
feature_dictionary_path = PROJECT_ROOT / CONFIG["paths"]["feature_dictionary"]

checks = pd.DataFrame(
    [
        {
            "artifact": "Validation predictions",
            "path": str(prediction_path),
            "exists": prediction_path.exists(),
        },
        {
            "artifact": "Best parameters",
            "path": str(best_parameters_path),
            "exists": best_parameters_path.exists(),
        },
        {
            "artifact": "Feature dictionary",
            "path": str(feature_dictionary_path),
            "exists": feature_dictionary_path.exists(),
        },
    ]
)
display(checks)

if prediction_path.exists():
    predictions = pd.read_parquet(prediction_path)
    display(predictions.head())
    print(predictions.columns.tolist())
    print(f"Rows: {len(predictions):,}")
```
