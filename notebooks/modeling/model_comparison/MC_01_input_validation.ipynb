# MC 01 — Input Validation

Verify that standardized outputs from every candidate branch are available before comparing models.

```python

from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd()
while not (PROJECT_ROOT / "src").exists() and PROJECT_ROOT != PROJECT_ROOT.parent:
    PROJECT_ROOT = PROJECT_ROOT.parent

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

CONFIG_PATH = PROJECT_ROOT / "configs" / "model_comparison.yaml"
```

```python
from src.ontario_peak_risk.model_comparison.io import (
    load_config,
    load_model_reports,
    validate_loaded_reports,
)

config, project_root = load_config(CONFIG_PATH)

fc_reports = load_model_reports(config, project_root, "forecasting")
pk_reports = load_model_reports(config, project_root, "peak_risk")

validation = __import__("pandas").concat(
    [
        validate_loaded_reports(fc_reports),
        validate_loaded_reports(pk_reports),
    ],
    ignore_index=True,
)

display(validation)
```

All models should have global, fold, FSA, and horizon metrics before the selection is approved. Runtime/tuning reports are optional evidence.
