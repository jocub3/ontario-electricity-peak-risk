## Selection rules

**Forecasting:** MAE is the primary metric. RMSE, MAPE, fold stability, horizon degradation, M9W-2023, and runtime evidence are guardrails.

**Peak-Risk:** PR-AUC is the primary metric. F1, Recall, Brier Score, fold stability, horizon behavior, M9W-2023, and runtime evidence are guardrails.

The 2025 holdout is not used in this phase.
