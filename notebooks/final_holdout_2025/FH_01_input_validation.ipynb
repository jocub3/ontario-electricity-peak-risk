# FH 01 — Input Validation

Verify the final split and both freeze manifests before running the one-time holdout.

```python
# Import libraries
from pathlib import Path
import sys
import yaml
import json
```

```python
# Define config paths
PROJECT_ROOT = Path.cwd()
while not (PROJECT_ROOT / "src").exists() and PROJECT_ROOT != PROJECT_ROOT.parent:
    PROJECT_ROOT = PROJECT_ROOT.parent
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))
CONFIG_PATH = PROJECT_ROOT / "configs" / "final_holdout_2025.yaml"
CONFIG_PATH
```

```python
# Get model's information
with CONFIG_PATH.open("r", encoding="utf-8") as h:
    cfg = yaml.safe_load(h)

for key in ["rf_freeze_manifest", "xgb_freeze_manifest", "modeling_foundation_config"]:
    path = PROJECT_ROOT / cfg["paths"][key]
    print(key, "->", path, "exists:", path.exists())

rf = json.loads((PROJECT_ROOT / cfg["paths"]["rf_freeze_manifest"]).read_text(encoding="utf-8"))
xgb = json.loads((PROJECT_ROOT / cfg["paths"]["xgb_freeze_manifest"]).read_text(encoding="utf-8"))

```

```python
print("RF frozen:", rf["hyperparameters"])
print("XGB frozen:", xgb["hyperparameters"])
print("XGB threshold:", xgb["operational_threshold"])
print("RF holdout previously evaluated:", rf["final_2025_holdout_evaluated"])
print("XGB holdout previously evaluated:", xgb["final_2025_holdout_evaluated"])
```
