Confirm that every expected serialized model and metadata file exists.

```python
# Import libraries
from pathlib import Path
import sys
import json
import pandas as pd
```

```python
# Define config paths
PROJECT_ROOT = Path.cwd()
while not (PROJECT_ROOT / "src").exists() and PROJECT_ROOT != PROJECT_ROOT.parent:
    PROJECT_ROOT = PROJECT_ROOT.parent
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))
CONFIG_PATH = PROJECT_ROOT / "configs" / "final_training.yaml"
CONFIG_PATH
```

```python
# Config 
ART = PROJECT_ROOT / "artifacts" / "final_models"

checks = []
for h in range(1, 25):
    checks.append(("rf", h, (ART / "forecasting_random_forest" / f"rf_h{h:02d}.joblib").exists()))
    checks.append(("xgb", h, (ART / "peak_risk_xgboost" / f"xgb_h{h:02d}.joblib").exists()))



```

```python
display(pd.DataFrame(checks, columns=["task", "horizon", "exists"]))
```
